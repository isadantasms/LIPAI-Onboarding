## 4. Regras rápidas para nomes de variáveis

Um nome de variável:

- não deve conter espaços;
- não deve começar com número;
- pode usar letras, números depois do primeiro caractere e `_`;
- diferencia maiúsculas de minúsculas.

## 5. Exemplo guiado: verificando tipos com `type()`

```python
quantidade = 10
preco = 12.5
curso = "Python"
ativo = True

print(type(quantidade))
print(type(preco))
print(type(curso))
print(type(ativo))
```

Uma saída como `<class 'int'>` indica que o valor pertence ao tipo `int`. Nesta lição, concentre-se no nome do tipo: `int`, `float`, `str` ou `bool`.

## 8. `print()` com vírgula e `str()`

```python
idade = 20

print("Idade:", idade)

mensagem = "Idade: " + str(idade)
print(mensagem)
```

`print("Idade:", idade)` funciona porque `print()` pode receber vários valores separados por vírgula.

Já em `"Idade: " + str(idade)`, `str()` converte o número para texto antes da junção com `+`.
