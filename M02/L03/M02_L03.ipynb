## 17. Reflexão final

Responda com suas palavras nas linhas abaixo.

**1. Qual é a diferença principal entre acessar um item de uma lista e acessar um valor de um dicionário?**

> A principal diferença é no "endereço" que usamos para encontrar a informação. Quando tentamos o encontrar pela *lista*, nós acessamos pela **posição numérica** (índice). Quando vamos acessar pelo *dicionário*, acessamos pela **chave** (com palavras ou "etiquetas" que criamos), então a ordem não importa muito.

**2. Quando um dicionário parece mais claro do que uma lista?**

> Um dicionário é mais claro quando precisamos agruoar várias características de um mesmo item e precisamos saber o que cada uma das "etiquetas" significam. Se usarmos uma lista com dois dados numéricos diferentes que se referem a duas coisas divergentes, vamos ter que adivinhar o que cada uma significa

**3. O que você deve verificar ao encontrar um `IndexError` ou um `KeyError`?**

> **Index Error**: Acontece quando tentamos acessar listas. Às vezes, acabamos nos confundindo e tentando acessar um índice numérico inexistente.

> **KeyError**: Acontece quando vamos usar dicionários. Precisamos checar se nossa "etiqueta" a ser acessada realmente tem aquele nome mesmo ou se estamos digitando diferente, pois ela precisa ser exatamente igual
