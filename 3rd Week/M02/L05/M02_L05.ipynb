## 12. `print()` e `return`

## 14. Exercício — função de classificação

Crie uma função chamada `classificar_nota` que receba uma nota e retorne:

- `"Aprovado"` se a nota for maior ou igual a `6`;
- `"Revisar"` se a nota for menor que `6`.

Exemplo de entrada:


## 15. Exemplo guiado: função com lista e `for`

````python
def contar_pares(numeros):
    quantidade = 0

    for numero in numeros:
        if numero % 2 == 0:
            quantidade = quantidade + 1

    return quantidade

valores = [2, 5, 8, 9, 10]
print(contar_pares(valores))
````

Observe que `return quantidade` aparece depois do `for`. A função precisa terminar de percorrer a lista antes de devolver a contagem.

## 16. Exercício — função com lista e contador

Crie uma função chamada `contar_maiores_que_10` que:

1. receba uma lista de números;
2. percorra a lista com `for`;
3. conte quantos valores são maiores que `10`;
4. retorne a quantidade.

Exemplo de entrada:

```python
[4, 12, 20, 7, 11]
```

Saída esperada:

```text
3
```

````python
# Complete aqui.
def contar_maiores_que_10(numeros):
  quantidade = 0

  for numero in numeros:
    if numero > 10:
      quantidade = quantidade + 1

  return quantidade

valores = [4, 12, 20, 7, 11]
print(contar_maiores_que_10(valores))
````

## 19. Desafio extra — opcional

Este desafio é para quem quiser praticar um pouco mais.

Considere:

```python
amostras = [
    {"id": "img_001", "classe": "A"},
    {"id": "img_002", "classe": "B"},
    {"id": "img_003", "classe": "A"}
]
```

Crie:

1. uma função `resumir_amostra(amostra)` que receba um dicionário e **retorne** uma frase com o identificador e a classe;
2. uma função `contar_classe(lista_amostras, classe_desejada)` que percorra a lista e **retorne** quantas amostras possuem a classe escolhida.

Exemplo de comportamento esperado para a primeira função:

```text
Amostra img_001 pertence à classe A
```

Exemplo de entrada para a segunda função:

```python
contar_classe(amostras, "A")
```

Saída esperada:

```text
2
```

````python
amostras = [
    {"id": "img_001", "classe": "A"},
    {"id": "img_002", "classe": "B"},
    {"id": "img_003", "classe": "A"}
]

# Desafio opcional: complete abaixo.
def resumir_amostra(amostra):
    return f"Amostra {amostra['id']} pertence à classe {amostra['classe']}"

def contar_classe(lista_amostras, classe_desejada):
    quantidade = 0
    for amostra in lista_amostras:
        if amostra["classe"] == classe_desejada:
            quantidade = quantidade + 1
    return quantidade

print(resumir_amostra(amostras[0]))

resultado = contar_classe(amostras, "A")
print(resultado)
````
