## 26. Erro adicional — separador incorreto

Considere um arquivo que usa `;`.

````python
csv_ponto_virgula = Path("dados_ponto_virgula.csv")

with open(csv_ponto_virgula, "w", encoding="utf-8") as f:
    f.write("nome;quantidade\n")
    f.write("classe_a;2\n")
````

````python
with open(csv_ponto_virgula, "r", encoding="utf-8", newline="") as f:
    leitor = csv.reader(f)
    for linha in leitor:
        print(linha)
````

Observe que, usando o separador padrão do leitor, a linha pode aparecer como **um único campo**.

Para este arquivo, o leitor pode receber `delimiter=";"`:

````python
with open(csv_ponto_virgula, "r", encoding="utf-8", newline="") as f:
    leitor = csv.reader(f, delimiter=";")
    for linha in leitor:
        print(linha)
````

O `catalogo.csv` deve conter:


Depois, crie uma função que leia o `catalogo.csv` **sem pandas** e mostre quantos arquivos existem em cada categoria.

Use apenas:

````python
# DESAFIO EXTRA — escreva sua solução aqui
import csv
from pathlib import Path

dataset_desafio = Path("dataset_desafio")
classe_a = dataset_desafio / "classe_a"
classe_b = dataset_desafio / "classe_b"

dataset_desafio.mkdir(exist_ok=True)
classe_a.mkdir(exist_ok=True)
classe_b.mkdir(exist_ok=True)

with open(classe_a / "item_001.txt", "w", encoding = "utf-8") as f:
    f.write("Amostra 1 da classe A\n")

with open(classe_a / "item_002.txt", "w", encoding = "utf-8") as f:
    f.write("Amostra 2 da classe A\n")

with open(classe_b / "item_003.txt", "w", encoding = "utf-8") as f:
    f.write("Amostra 3 da classe B\n")

with open(classe_b / "item_004.txt", "w", encoding = "utf-8") as f:
    f.write("Amostra 4 da classe B\n")

catalogo_desafio = dataset_desafio / "catalogo.csv"
colunas = ["arquivo", "categoria"]

registros = [
    {"arquivo": "item_001.txt", "categoria": "classe_a"},
    {"arquivo": "item_002.txt", "categoria": "classe_a"},
    {"arquivo": "item_003.txt", "categoria": "classe_b"},
    {"arquivo": "item_004.txt", "categoria": "classe_b"}
]

with open(catalogo_desafio, "w", encoding = "utf-8", newline="") as f:
    escritor = csv.DictWriter(f, fieldnames = colunas)
    escritor.writeheader()
    escritor.writerows(registros)

def contar_por_categoria(caminho_csv):
    contagem = {}
    with open(caminho_csv, "r", encoding="utf-8", newline="") as f:
        leitor = csv.DictReader(f)
        for linha in leitor:
            if linha["categoria"] not in contagem:
                contagem[linha["categoria"]] = 1
            else:
                contagem[linha["categoria"]] += 1
    return contagem

resultado = contar_por_categoria(catalogo_desafio)
print(resultado)
````
