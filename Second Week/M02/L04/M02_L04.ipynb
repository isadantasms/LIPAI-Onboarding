## 5. `if`, `elif` e `else`

Python testa as condições de cima para baixo. Quando encontra a primeira condição verdadeira, executa aquele bloco.

````python
nota = 7

if nota >= 9:
    print("Excelente")
elif nota >= 6:
    print("Aprovado")
else:
    print("Precisa revisar")
````

## 6. Indentação: dentro e depois do bloco

No primeiro exemplo, duas linhas estão dentro do `if`. A última está depois do bloco.

````python
idade = 20

if idade >= 18:
    print("Condição verdadeira")
    print("Esta linha está dentro do if")

print("Esta linha está depois do if")
````

### Exercício 5 — Reorganize a indentação

O trecho abaixo representa a intenção: mostrar duas mensagens **somente quando** a nota for maior ou igual a 6, e depois mostrar `"Fim"` independentemente da nota.

```text
nota = 7

if nota >= 6:
print("Aprovado")
print("Pode continuar")

    print("Fim")
```

Na célula seguinte, reescreva o código com a indentação correta. Depois, explique em uma frase por que a indentação altera o fluxo.

````python
nota = 7

# TODO: reescreva o código com a indentação correta.

if nota >= 6:
  print("Aprovado")
  print("Pode continuar")

print("Fim")
````

**Minha explicação sobre a indentação:** Se a linha de código estiver 'fora do bloco' (ou seja, sem o alinhamento correto), o programa simplesmente não roda. A indenização altera o fluxo por ser ela que organiza quem está dentro do bloco if e quem está de fora. Se houver desalinhamento, o programa não reconhece quem faz o quê.

## 7. Repetição com `for`

Antes de executar, acompanhe mentalmente:

## 8. `for` e `range()`

Compare os dois exemplos.

````python
nomes = ["Ana", "Bruno", "Carla"]

print("Percorrendo os itens da lista:")
for nome in nomes:
    print(nome)

print("Percorrendo números produzidos por range:")
for numero in range(3):
    print(numero)
````

### Exercício 7 — Quando usar cada forma?

Responda em texto:

1. Se você quer apenas imprimir cada nome de uma lista, por que `for nome in nomes` é mais direto do que gerar índices com `range()`?
2. Complete um exemplo que use `range()` para mostrar os números de 1 a 5.
3. Em uma frase, explique a diferença entre o papel de `for` e o de `range()`.

**Resposta 1:** Usar um `for nome in nomes`é mais direta pois acessa os valores da lista automaticamente. Já utilizando o `range()`, precisaríamos criar um índice, calcular o tamanho da lista e então, por fim, escrever um código para buscar o item pela sua posição, sendo muito mais trabalhoso.

**Resposta 3:** O `for`é o comando que cria o laço de repetição, o que faz nosso código rodar várias vezes, enquanto o `range()`é a função que gera a sequência de números que será percorrida pelo `for`

````python
# TODO: use range() junto com for para mostrar os números de 1 a 5.

for numero in range(1, 6):
  print(numero)
````

## 9. Contador

Um contador começa com um valor e é atualizado quando algo acontece.

````python
idades = [15, 22, 18, 14]
quantidade = 0

for idade in idades:
    if idade >= 18:
        quantidade = quantidade + 1

print("Quantidade de maiores de idade:", quantidade)
````

### Exercício 8 — Conte itens

Conte quantos valores da lista são maiores que 10.

````python
valores = [4, 12, 20, 7, 15]
quantidade = 0

# TODO: use for + if e atualize quantidade.
# Ao final, mostre o valor de quantidade.

for valor in valores:
  if valor > 10:
    quantidade = quantidade + 1

print("Quantidade de valores maiores que 10:", quantidade)
````

## 14. Diagnóstico rápido: `NameError` e `TypeError`

Sem executar os trechos incorretos, identifique o problema em cada caso.

**Caso A**

```text
idade = 20
print(idades)
```

**Caso B**

```text
idade = "20"

```

Escreva:

- qual erro você espera em cada caso;
- qual detalhe do código ajuda a perceber o problema.

> **Caso B:** `TypeError`, pois o número `"20"`está entre aspas na primeira linha, o que faz o programa entendê-lo como `str`. Quando o código tenta somar com um inteiro `1`, o programa "bloqueia" isso porque mistura texto e número.

## 15. Desafio extra — opcional

Crie uma pequena lista de dicionários representando amostras. Cada amostra deve ter:

- `id`;
- `classe`;
- `medida`.

Depois:

1. escolha um limite;
2. percorra a lista com `for`;
3. use `if` para verificar a medida;
4. conte quantas amostras estão acima do limite.

Este desafio é opcional.

````python
# TODO opcional: construa sua lista, escolha um limite e crie o contador.

amostras = [
    {"id": "A001", "classe": "classe_a", "medida": 12.5},
    {"id": "A002", "classe": "classe_b", "medida": 8.0},
    {"id": "A003", "classe": "classe_a", "medida": 15.3},
    {"id": "A004", "classe": "classe_c", "medida": 9.5},
    {"id": "A005", "classe": "classe_c", "medida": 11.2}
]

limite = 10.5

contador = 0

for amostra in amostras:
    if amostra["medida"] > limite:
        contador = contador + 1

print("Total de amostras acima do limite:", contador)
````
